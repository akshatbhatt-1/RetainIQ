# Advanced Feature Engineering

```python
train_df["income_per_dependent"] = (
    train_df["annual_income"] /
    (train_df["dependent_count"] + 1)
)

train_df["products_per_tenure"] = (
    train_df["number_of_products"] /
    (train_df["tenure_months"] + 1)
)

train_df["engagement_score"] = (
    train_df["digital_engagement_index"] *
    train_df["total_digital_logins"]
)

train_df["relationship_depth_score"] = (
    train_df["number_of_products"] *
    train_df["tenure_months"]
)

train_df["high_value_customer"] = (
    train_df["customer_lifetime_value"] >
    train_df["customer_lifetime_value"].median()
).astype(int)

train_df["inactive_high_value"] = (
    (
        train_df["high_value_customer"] == 1
    ) &
    (
        train_df["digital_engagement_index"] <
        train_df["digital_engagement_index"].median()
    )
).astype(int)

train_df["financial_stress_score"] = (
    train_df["debt_to_income_ratio"] *
    (
        1 /
        (train_df["annual_income"] + 1)
    )
)

train_df["loyalty_index"] = (
    train_df["tenure_months"] *
    train_df["loyalty_program_member"]
)

train_df.head()
```

# Train Validation Split

```python
from sklearn.model_selection import train_test_split

possible_leakage = [
    "retention_offer_received",
    "retention_offer_accepted",
    "discount_or_fee_waiver_received",
    "unresolved_complaint_count",
    "complaint_resolution_time",
    "last_campaign_response_days",
    "service_request_count",
    "escalation_count",
    "account_inactive_days",
    "last_login_days"
]

X = train_df.drop(
    columns=[target_col] + possible_leakage + id_cols + drop_cols
)

y = train_df[target_col]

X_train, X_valid, y_train, y_valid = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print(X_train.shape)
print(X_valid.shape)
```

```python
categorical_cols = X_train.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_cols = X_train.select_dtypes(
    exclude=["object"]
).columns.tolist()

print(len(categorical_cols))
print(len(numerical_cols))
```

# Preprocessing Pipeline

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numerical_cols
        ),
        (
            "cat",
            categorical_transformer,
            categorical_cols
        )
    ]
)
```

# Baseline Logistic Regression Model

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

baseline_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "classifier",
            LogisticRegression(
                max_iter=1000,
                class_weight="balanced",
                random_state=42
            )
        )
    ]
)

baseline_model.fit(
    X_train,
    y_train
)
```

```python
baseline_preds = baseline_model.predict(
    X_valid
)

baseline_probs = baseline_model.predict_proba(
    X_valid
)[:, 1]
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

print(
    classification_report(
        y_valid,
        baseline_preds
    )
)

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_valid,
        baseline_preds
    )
)

print(
    "\nROC-AUC:",
    roc_auc_score(
        y_valid,
        baseline_probs
    )
)

print(
    "\nPR-AUC:",
    average_precision_score(
        y_valid,
        baseline_probs
    )
)
```
